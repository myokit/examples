# Technical note: Variable types

This note will discuss Myokit's division of variables into `state`, `intermediary`, `constant`, and `bound`.



In Myokit, "constant" has the slightly odd meaning "does not (directly or indirectly) depend on a statem state derivative, or input (bound variable)".

```python
m = myokit.parse_model('''
[[model]]
x.a = 0

[x]
t = 0 bind time
dot(a) = 1
b = dot(a)
c = 2
d = 1 + c
''')

print('a', m.get('x.a').is_constant())  # False, a changes in time
print('b', m.get('x.b').is_constant())  # Debatably false: depends on state RHS, assumed variable
print('c', m.get('x.c').is_constant())  # True, equals 2
print('d', m.get('x.d').is_constant())  # True, depends on a constant
print('t', m.get('x.t').is_constant())  # False, is an external input
print()
print('a rhs', m.get('x.a').rhs().is_constant())  # True, equals 1
print('b rhs', m.get('x.b').rhs().is_constant())  # Debatably false, depends on dot()
print('c rhs', m.get('x.c').rhs().is_constant())  # True
print('d rhs', m.get('x.d').rhs().is_constant())  # True
print('t rhs', m.get('x.t').rhs().is_constant())  # True

```
